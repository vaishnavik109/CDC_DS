```python
import os
import pandas as pd

def verify_images_from_drive(train_dir, test_dir, save_path):
    records = []


    train_images = [f for f in os.listdir(train_dir) if f.endswith(".png")]
    for img in train_images:
        records.append({
            "split": "train",
            "image_name": img,
            "exists": os.path.exists(os.path.join(train_dir, img))
        })


    test_images = [f for f in os.listdir(test_dir) if f.endswith(".png")]
    for img in test_images:
        records.append({
            "split": "test",
            "image_name": img,
            "exists": os.path.exists(os.path.join(test_dir, img))
        })

    df = pd.DataFrame(records)


    train_completion = df[df.split == "train"]["exists"].mean() * 100
    test_completion  = df[df.split == "test"]["exists"].mean() * 100
    overall_completion = df["exists"].mean() * 100

    print(f"Train images completion: {train_completion:.2f}% ({len(train_images)})")
    print(f"Test images completion:  {test_completion:.2f}% ({len(test_images)})")
    print(f"Overall completion:     {overall_completion:.2f}%")


    df.to_csv(save_path, index=False)
    print(f"Manifest saved to: {save_path}")

    return df
```

```python
import os
import pandas as pd

def verify_images_from_drive(image_dir, split_name):
    images = [f for f in os.listdir(image_dir) if f.endswith(".png")]

    df = pd.DataFrame({"image_name": images})
    df["exists"] = df["image_name"].apply(
        lambda x: os.path.exists(os.path.join(image_dir, x))
    )

    completion = df["exists"].mean() * 100
    print(f"{split_name} image completion: {completion:.2f}% ({len(images)} images)")

    return df
```

```python
train_df = verify_images_from_drive(IMAGE_TRAIN_DIR, "Train")
test_df  = verify_images_from_drive(IMAGE_TEST_DIR, "Test")

manifest = pd.concat([train_df.assign(split="train"),
                       test_df.assign(split="test")])

manifest.to_csv("/content/drive/MyDrive/image_manifest.csv", index=False)
```

```python
import torch
import torchvision.models as models
from torchvision import transforms
from PIL import Image
import os

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

```

```python
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
resnet = models.resnet18(pretrained=True)
resnet.fc = torch.nn.Identity()
resnet = resnet.to(device)
resnet.eval()
```

```python
def extract_image_embedding(image_name):
    path = os.path.join(IMAGE_TRAIN_DIR, image_name)

    img = Image.open(path).convert("RGB")
    img = transform(img).unsqueeze(0).to(device)

    with torch.no_grad():
        emb = resnet(img)

    return emb.cpu().numpy().flatten()
```

```python
import numpy as np

X_tab_img = train_img.drop(columns=["price", "image_name", "image_exists"])
y_img = np.log1p(train_img["price"])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X_tr_tab, X_val_tab, y_tr, y_val, idx_tr, idx_val = train_test_split(
    X_tab_img,
    y_img,
    train_img.index,
    test_size=0.2,
    random_state=42
)

X_tr_img = X_img[idx_tr]
X_val_img = X_img[idx_val]
```

```python
scaler = StandardScaler()

X_tr_img = scaler.fit_transform(X_tr_img)
X_val_img = scaler.transform(X_val_img)
```

```python
X_tr_fusion = np.hstack([X_tr_tab.values, X_tr_img])
X_val_fusion = np.hstack([X_val_tab.values, X_val_img])

print("Fusion train shape:", X_tr_fusion.shape)
print("Fusion val shape:", X_val_fusion.shape)
```

```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, r2_score

xgb_fusion = XGBRegressor(
    n_estimators=600,
    learning_rate=0.03,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    n_jobs=-1
)

xgb_fusion.fit(X_tr_fusion, y_tr)
```

```python
fusion_preds = xgb_fusion.predict(X_val_fusion)

rmse_fusion = np.sqrt(mean_squared_error(y_val, fusion_preds))
r2_fusion = r2_score(y_val, fusion_preds)

print(" Multimodal Fusion RMSE:", rmse_fusion)
print(" Multimodal Fusion R²:", r2_fusion)
```

```python
final_price_preds = np.expm1(fusion_preds)
```
